```python
import pandas as pd
import glob

# List all HDF5 files
hdf5_files = glob.glob('/content/Data-HDF5/*.hdf5')
print(f"Found {len(hdf5_files)} files.")

# Columns to keep
columns_to_keep = ['EMG_17', 'EMG_18', 'EMG_19', 'EMG_20',
                   'EMG_21', 'EMG_22', 'EMG_23', 'EMG_24',
                   'TRAJ_1', 'TRAJ_GT']

# Read only selected columns from each file, then concatenate
data = pd.concat([pd.read_hdf(f)[columns_to_keep] for f in hdf5_files], ignore_index=True)

# Show a sample
display(data.head(10))
```

```python
import numpy as np

# Only keep labeled gestures, skip -1.0 label (uncertain/no gesture)
filtered = data[data['TRAJ_GT'] != -1.0].reset_index(drop=True)
print("\nLabel counts after filtering uncertain/no-gesture (-1.0):")
print(filtered['TRAJ_GT'].value_counts())

# Parameters
channels = [f'EMG_{i}' for i in range(17, 25)]  # Wrist channels
window_size = 200  # samples, e.g. ~40ms at 5120Hz
stride = 100       # overlap window by 50%

X_features = []
y_labels = []

for start_idx in range(0, len(filtered) - window_size, stride):
    window = filtered.iloc[start_idx : start_idx + window_size]
    # Compute RMS feature for each channel in this window
    rms_features = window[channels].apply(lambda arr: np.sqrt(np.mean(arr**2)), axis=0)
    X_features.append(rms_features.values)
    # Take the most frequent label in this window as the target
    y_labels.append(window['TRAJ_GT'].mode()[0])

# Convert to arrays
X = np.stack(X_features)
y = np.array(y_labels)

print(f"Shape of features: {X.shape} (windows, channels)")
print(f"Shape of labels: {y.shape} (windows,)")

# Show new label distribution
from collections import Counter
print("Label counts after windowing:")
print(Counter(y))
```

```python
import numpy as np
from collections import Counter
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import f1_score, accuracy_score
from sklearn.model_selection import train_test_split
import pandas as pd

# --- Balance Dataset ---
idle_idx = np.where(y == 0.0)[0]
other_idx = np.where(y != 0.0)[0]
n_sample = min([np.sum(y == val) for val in set(y) if val != 0.0])
np.random.seed(42)
idle_idx_sample = np.random.choice(idle_idx, size=n_sample, replace=False)
balanced_indices = np.concatenate([idle_idx_sample, other_idx])

X_bal = X[balanced_indices]
y_bal = y[balanced_indices]
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.callbacks import EarlyStopping

early_stopping = EarlyStopping(
    monitor='val_loss',    # Monitor validation loss
    patience=15,           # Number of epochs with no improvement after which training will be stopped
    restore_best_weights=True,  # Restore model weights from the epoch with the best value of the monitored quantity
    verbose=1
)

# Your original data: X, y
# ------------------------

# Balance dataset as in your sklearn code
idle_idx = np.where(y == 0.0)[0]
other_idx = np.where(y != 0.0)[0]
n_sample = min([np.sum(y == val) for val in set(y) if val != 0.0])
np.random.seed(42)
idle_idx_sample = np.random.choice(idle_idx, size=n_sample, replace=False)
balanced_indices = np.concatenate([idle_idx_sample, other_idx])

X_bal = X[balanced_indices]
y_bal = y[balanced_indices]

# Scale features for Keras training
scaler = StandardScaler()
X_bal_scaled = scaler.fit_transform(X_bal)

# Remap non-continuous labels to contiguous 0..num_classes-1 for Keras
unique_labels = np.unique(y_bal)
label_map = {label: idx for idx, label in enumerate(unique_labels)}
print('Label mapping:', label_map)
y_bal_mapped = np.array([label_map[label] for label in y_bal])

# Train/test split with mapped labels
X_train, X_test, y_train, y_test = train_test_split(
    X_bal_scaled, y_bal_mapped, test_size=0.2, stratify=y_bal_mapped, random_state=42)

print('Train X shape:', X_train.shape)
print('Test X shape:', X_test.shape)

input_dim = X_train.shape[1]
num_classes = len(unique_labels)
print('Number of classes:', num_classes)

# Build Keras model
model = models.Sequential([
    layers.Input(shape=(input_dim,)),
    layers.Dense(100, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train model with validation split and verbose output
history = model.fit(
    X_train, y_train,
    epochs=200,
    batch_size=32,
    validation_split=0.2,
    verbose=2,
    callbacks=[early_stopping]
)

# Evaluate on test set
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=2)
print(f'Test accuracy: {test_acc:.4f}')
```

```python
import tensorflow as tf

# Convert the trained Keras model to TensorFlow Lite format
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()

# Save the TensorFlow Lite model to a file
with open('mlp_model.tflite', 'wb') as f:
    f.write(tflite_model)

print("TensorFlow Lite model saved as 'mlp_model.tflite'")
```
